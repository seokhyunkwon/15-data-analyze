```python
# 1)

import pandas as pd

student_scores = pd.Series([85, 92, 78, 90, 88])
print(student_scores)
print()

student_scores.index = ['Alice', 'Bob', 'Charlie', 'David', 'Eve']
print(student_scores)
print()

exam_scores = pd.Series({
    'Python': 95,
    'SQL': 85,
    'Web': 90
})

print(exam_scores)
```

```python
# 2)

charlie_score = student_scores['Charlie']
print(charlie_score)
print()

selected_scores = student_scores.iloc[1:4]
print(selected_scores)
print()
# 점수만 출력이라면
only_scores = student_scores.values[1:4]
print(only_scores)
```
